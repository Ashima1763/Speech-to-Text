```python
import torch
import torchaudio
from transformers import pipeline
from pathlib import Path
```

```python
# Whisper model from Hugging Face
pipe = pipeline(model="openai/whisper-base")
```

```python
# test MP3 audio file
audio_path = "/content/test audio.opus"
```

```python
result = pipe(audio_path)
print("\nTranscription:\n")
print(result["text"])
```
